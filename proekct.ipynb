### 3. Создание профиля пользователя

В будущем отдел маркетинга планирует создать модель для прогнозирования возврата пользователей. Поэтому сейчас они просят вас построить агрегированные признаки, описывающие поведение и профиль каждого пользователя.

После этого добавьте два бинарных признака:

**Рекомендация:** перед тем как строить профиль, отсортируйте данные по времени совершения заказа.

```python
import pandas as pd

# Убедимся, что работаем с очищенными данными
df = df_final.copy()

# --- 1. Сортировка по времени заказа (обязательно по условию) ---
df = df.sort_values(by='order_ts', ascending=True)

# --- 2. Признаки первого заказа (по user_id) ---
first_order = (
    df.groupby('user_id')
      .agg(
          first_order_dt=('order_dt', 'first'),
          first_device=('device_type_canonical', 'first'),
          first_region=('region_name', 'first'),
          first_partner=('service_name', 'first'),
          first_event_type=('event_type_main', 'first')
      )
      .reset_index()
)

# --- 3. Агрегации по пользователю (общие метрики) ---
user_profile = (
    df.groupby('user_id')
      .agg(
          last_order_dt=('order_dt', 'last'),
          total_orders=('order_id', 'count'),
          avg_revenue_rub=('revenue_rub', 'mean'),
          avg_tickets_count=('tickets_count', 'mean')
      )
      .reset_index()
)

# --- 4. Среднее время между заказами (days_since_prev) ---
# days_since_prev уже посчитан в исходных данных; среднее по нему — это среднее время между заказами
avg_time_between_orders = (
    df.groupby('user_id')['days_since_prev']
      .mean()
      .rename('avg_days_between_orders')
      .reset_index()
)

# Объединяем все части профиля
user_profile = (
    user_profile
    .merge(first_order, on='user_id', how='left')
    .merge(avg_time_between_orders, on='user_id', how='left')
)

# --- 5. Бинарные признаки ---
user_profile['is_two'] = (user_profile['total_orders'] >= 2).astype(int)
user_profile['is_five'] = (user_profile['total_orders'] >= 5).astype(int)

# --- 6. Проверка результата ---
print("Размер профиля пользователя:", user_profile.shape)
print("\nПервые 10 строк профиля:")
print(user_profile.head(10))

print("\nСтатистика по total_orders:")
print(user_profile['total_orders'].describe())

print("\nДоли пользователей с 2+ и 5+ заказами:")
print("Доля is_two:", user_profile['is_two'].mean())
print("Доля is_five:", user_profile['is_five'].mean())
```

- На этапе 3.1 построен профиль пользователя на основе очищенных данных. Для каждого пользователя рассчитаны дата первого и последнего заказа, характеристики первого заказа (устройство, регион, партнёр, жанр), а также агрегированные метрики: общее количество заказов, средний чек, среднее число билетов и среднее время между заказами. Добавлены бинарные признаки is_two и is_five.

- Профиль содержит 21 821 пользователя. Медианное количество заказов — 2, 75‑й перцентиль — 5, среднее сильно завышено (13.23) из‑за выбросов. Доля пользователей с 2+ заказами — 61.8 %, с 5+ заказами — 29.1 %.

- Обнаружен пользователь с 10 256 заказами — требуется отдельная проверка на техническую природу. Пропуски в признаке «среднее время между заказами» соответствуют пользователям с единственным заказом и должны быть обработаны перед моделированием.

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

# Функция для расчёта доли возврата и построения графика (топ-10 по размеру сегмента)
def plot_return_rate(df, col, name, top_n=10):
    # Агрегация: кол-во пользователей и доля тех, у кого is_two == 1
    agg = (
        df.groupby(col, dropna=False)
          .agg(
              users_count=('user_id', 'nunique'),
              return_rate=('is_two', 'mean')
          )
          .reset_index()
    )
    
    # Сортировка по количеству пользователей (чтобы взять топ-сегменты по объёму)
    agg = agg.sort_values('users_count', ascending=False)
    top_segments = agg.head(top_n)
    
    avg_return_rate = df['is_two'].mean()
    
    fig, ax = plt.subplots(figsize=(10, 6))
    bars = ax.bar(top_segments[col], top_segments['return_rate'], color='#4C72B0')
    ax.axhline(avg_return_rate, color='red', linestyle='--', linewidth=2, label=f'Средняя доля возврата ({avg_return_rate:.2%})')
    
    ax.set_title(f'Доля пользователей с 2+ заказами по сегментам: {name}', fontsize=14)
    ax.set_xlabel(name, fontsize=12)
    ax.set_ylabel('Доля возврата (is_two)', fontsize=12)
    ax.tick_params(axis='x', rotation=45)
    ax.legend()
    
    # Подписи значений на столбцах
    for bar, rate in zip(bars, top_segments['return_rate']):
        height = bar.get_height()
        ax.text(bar.get_x() + bar.get_width()/2., height,
                f'{rate:.1%}',
                ha='center', va='bottom', fontsize=10)
    
    plt.tight_layout()
    plt.show()
    
    print(f"=== Статистика по {name} (топ-{top_n} сегментов по объёму) ===")
    print(top_segments[['users_count', 'return_rate']].round(3))
    print(f"Средняя доля возврата по всей выборке: {avg_return_rate:.3f}")
    return top_segments

# Строим графики для всех 4 признаков
plot_return_rate(df_filtered, 'first_event_type', 'тип первого мероприятия')
plot_return_rate(df_filtered, 'first_device', 'тип устройства первого заказа')
plot_return_rate(df_filtered, 'first_region', 'регион первого заказа', top_n=10)
plot_return_rate(df_filtered, 'first_partner', 'билетный оператор первого заказа', top_n=10)
```

- Отдельные сегменты (например, некоторые партнёры и регионы) показывают возврат выше среднего. Но важно учитывать размер сегмента: высокая доля возврата в малых группах (десятки пользователей) может быть случайной и не гарантировать такой же эффект при масштабировании.
- При интерпретации делали упор на крупные сегменты (тысячи пользователей) как на наиболее надёжные. Повышенные доли возврата в небольших сегментах отмечали как гипотезы, требующие дополнительного подтверждения.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

# Создаём сегмент по avg_tickets_count
def assign_ticket_segment(x):
    if 1 <= x < 2:
        return '1–2 билета'
    elif 2 <= x < 3:
        return '2–3 билета'
    elif 3 <= x < 5:
        return '3–5 билетов'
    else:  # 5 и более
        return '5+ билетов'

user_profile['ticket_segment'] = user_profile['avg_tickets_count'].apply(assign_ticket_segment)

# Агрегируем: количество пользователей и доля с повторными заказами (is_two == 1)
segmentation = (
    user_profile
    .groupby('ticket_segment', observed=True)
    .agg(
        users_count=('user_id', 'size'),
        return_rate=('is_two', 'mean'),          # доля пользователей с 2+ заказами
        median_tickets=('avg_tickets_count', 'median'),
        mean_tickets=('avg_tickets_count', 'mean')
    )
    .reset_index()
    .sort_values('ticket_segment')
)

print(segmentation)

# Визуализация: столбцы с долей повторных покупок
plt.figure(figsize=(10, 6))
sns.barplot(
    data=segmentation,
    x='ticket_segment',
    y='return_rate',
    palette='viridis'
)
plt.xlabel('Сегмент по среднему количеству билетов в заказе')
plt.ylabel('Доля пользователей с повторными заказами')
plt.title('Доля повторных заказов по сегментам количества билетов')
plt.ylim(0, 1)  # чтобы шкала была от 0 до 1
plt.show()
```

- Мы проанализировали влияние среднего количества билетов в заказе на вероятность повторной покупки, разделив пользователей на сегменты: «1–2 билета», «2–3 билета», «3–5 билетов», «5+ билетов».
- Распределение пользователей по сегментам неравномерно: основная доля приходится на сегменты «2–3 билета» (9 632 пользователя) и «3–5 билетов» (9 073 пользователя). Сегмент «5+ билетов» — наименьший по численности (684 пользователя).
- Доля повторных заказов существенно различается между сегментами. Наиболее высокая лояльность наблюдается в сегменте «2–3 билета»: 73,9 % пользователей совершают повторные покупки. Это может указывать на то, что сценарий покупки билетов на небольшую компанию способствует формированию привычки пользоваться сервисом.
- В сегменте «5+ билетов» доля повторных заказов аномально низкая — всего 19,6 %. Вероятно, такие заказы носят эпизодический или организационный характер и не отражают регулярную вовлечённость пользователя в сервис.
- Таким образом, количество билетов в заказе — значимый фактор сегментации пользователей: он позволяет выделить группу с высокой лояльностью («2–3 билета») и группу с низкой вовлечённостью, несмотря на крупный размер заказа («5+ билетов»). Эти сегменты стоит учитывать при формировании маркетинговых механик: для сегмента «2–3 билета» можно усиливать программы удержания и рекомендации, а для сегмента «5+ билетов» — прорабатывать отдельные сценарии вовлечения после разовой покупки (например, напоминания о новых крупных событиях или предложения для организаторов).

Это говорит о том, что лояльность формируется за счёт других факторов — например, удобства сервиса, релевантности рекомендаций или регулярности появления интересных мероприятий, а не за счёт дня, когда произошла первая покупка.
